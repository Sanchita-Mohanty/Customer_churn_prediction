DATA CLEANING

We need to convert the object data type to either float or int because we cant identify the nan values

```python
telecom_data_set=telecom_data.copy()
```

```python
telecom_data_set.TotalCharges=pd.to_numeric(telecom_data_set.TotalCharges,errors='coerce')
#error=coerence identifies if any value that cant be converted to int is replaced by nan
telecom_data_set.isnull().sum()
```

In TotalCharges 11 values are null vales

```python
telecom_data_set.loc[telecom_data_set['TotalCharges'].isnull()==True]
```

These 11 records can be out of 7043 records  as they are vey low in no. compared to size od dataset

```python
telecom_data_set.dropna(how='any',inplace=True)
```

Divided customers into bind based on tenure of <12 months

```python
print(telecom_data_set['tenure'].max())
```

Group tenure into bins

```python
labels=["{0} - {1}".format(i,i+11) for i in range(1,72,12)]
telecom_data_set['tenure_group']=pd.cut(telecom_data_set.tenure,range(1,80,12),right=False,labels=labels)
```

Remove Columns that are not required

Convet churn into numerical variable yes=1 no=0

```python
telecom_data_set['Churn']=np.where(telecom_data_set.Churn=='Yes',1,0)
```

```python
telecom_data_set.head()
```

Convert Categorical Variable into Numerical Variable

We can covert Categorical Variable into Numerical Variable by-
1.Label Encoding(assigning 1, 0,3.. based on their value but it can be misleading to the model,It should be used for target variable ot Y variables)
2.One Hot Encoding: Converts categorical variables into binary columns.
3.Dummy Trap: Occurs when all dummy variables are included, causing multi-collinearity.

```python
telecom_data_dummies=pd.get_dummies(telecom_data_set)
telecom_data_dummies = telecom_data_dummies.astype(int)
telecom_data_dummies.head()
```

Relationship between Month and Total Charges

```python
sns.lmplot(data=telecom_data_dummies,x='MonthlyCharges',y='TotalCharges',fit_reg=False)
sns.despine
```

Total Charges increases as the Monthly Charges Increases

Churn by Monthly Charges and Total Charges

```python
Monthly=sns.kdeplot(telecom_data_dummies.MonthlyCharges[(telecom_data_dummies['Churn']==0)],color='Black',shade=True)
Monthly=sns.kdeplot(telecom_data_dummies.MonthlyCharges[(telecom_data_dummies['Churn']==1)],ax=Monthly,color='Red',shade=True)
Monthly.legend(["No Churn","Churn"],loc='upper right')
Monthly.set_ylabel('Density')
Monthly.set_xlabel('Monthly Charges')
Monthly.set_title('Monthly charges by churn')
```

Churn is more in Higher Monthly Charges

```python
Tot = sns.kdeplot(telecom_data_dummies.TotalCharges[(telecom_data_dummies["Churn"] == 0) ],
                color="Black", shade = True)
Tot = sns.kdeplot(telecom_data_dummies.TotalCharges[(telecom_data_dummies["Churn"] == 1) ],
                ax =Tot, color="Red", shade= True)
Tot.legend(["No Churn","Churn"],loc='upper right')
Tot.set_ylabel('Density')
Tot.set_xlabel('Total Charges')
Tot.set_title('Total charges by churn')
```

Higher Churn rate at low Total Charges

Higher Monthly Charge, Lower tenure and Lower Total Charge are linkd to High Churn.

```python
non_churner=telecom_data_set.loc[telecom_data_set['Churn']==0]
churner=telecom_data_set.loc[telecom_data_set['Churn']==1]
```

```python
def uniplot(df, col, title, hue=None):
    sns.set_style('whitegrid')
    sns.set_context('talk')
    plt.rcParams["axes.labelsize"] = 20
    plt.rcParams['axes.titlesize'] = 22
    plt.rcParams['axes.titlepad'] = 30
    
    temp = pd.Series(data=hue)
    fig, ax = plt.subplots()
    width = len(df[col].unique()) + 7 + 4 * len(temp.unique())
    fig.set_size_inches(width, 8)
   

    # Remove log scale if counts are similar
    plt.yscale('linear')
    
    # Create count plot
    ax = sns.countplot(data=df, x=col, order=df[col].value_counts().index, hue=hue, palette='bright')
    
    # Set title
    plt.title(title)
    
    plt.show()
```

```python
uniplot(churner,col='Partner',title='Distribution of Gender for Churned Customers',hue='gender')
```

```python
uniplot(non_churner,col='Partner',title='Distribution of Gender for Non Churned Customers',hue='gender')
```

```python
uniplot(churner,col='SeniorCitizen',title='Distribution of SeniorCitizen for Churned Customers',hue='gender')
```
